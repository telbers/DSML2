# Milestone 1 – Citi Bike Boston: Prediction of usage

**Datensatz**: [Citi Bike Boston Trip Data](https://s3.amazonaws.com/hubway-data/202408-bluebikes-tripdata.zip), Januar 2023
**Prediction Task**: In this Project we are trying to predict the development of the usage from electronic and classics Bikes in the area of Boston. We will be looking at different years of months an compare the data to understand the development over the last few years and with that develop an prediction model for the future usage of bikes. We also want to include weather data as an additional feature to improve the data. We do this because we think there is a correlation between the weather an the willingsness to use bikes. As for the predictive Model we choose the regression Model.

The next steps will include the data preparation. Our goal here will be to sort the Data of the days of the month in time boxes for example morning, noon, afternoon, evening etc. With this we can accumulate the highest average of use-time through the day as well as highest and lowest peaks. 
**Geplante Features**:
- Starting time of bike-trip (`started_at`)
- Day of the week (`weekday`)
- Time period during the day (`time_of_day`)
- Average usage during time-periods ('average usage')
- highest usage during time-periods ('peak usage')
- lowest usage during time-periods ('lowest usage')
- Stunde (`hour`)
- Start station of trip (`start_station_id`)
- User-type (`member_casual`)
- Weather during trip (optional: Temperatur, Regen)

This project uses data from august 2023 with ca. 100.000 bike-trips.

```
import pandas as pd

# CSV-Datei einlesen
csv_path = r"C:\Users\hanac\PyCharmMiscProject\202408-bluebikes-tripdata.csv"
df = pd.read_csv(csv_path, engine="python", sep=None, on_bad_lines="warn")

print("🚩 Ursprünglicher Datensatz:")
print(df.shape)

# 1. Zeitfelder im deutschen Format korrekt parsen
df['started_at'] = pd.to_datetime(df['started_at'], dayfirst=True, errors='coerce')
df['ended_at'] = pd.to_datetime(df['ended_at'], dayfirst=True, errors='coerce')
print("✅ Nach datetime-Konvertierung:")
print(df.shape)

# 2. Zeilen mit ungültigem Datum entfernen
df = df.dropna(subset=['started_at', 'ended_at'])
print("✅ Nach Entfernen von NaT-Zeilen:")
print(df.shape)

# 3. Fahrtzeit in Sekunden berechnen
df['tripduration'] = (df['ended_at'] - df['started_at']).dt.total_seconds()

# 4. Nur sinnvolle (positive) Fahrten behalten
df = df[df['tripduration'] > 0]
print("✅ Nach Entfernen negativer/0-Fahrzeiten:")
print(df.shape)

# 5. Sicherstellen, dass tripduration numerisch ist
df['tripduration'] = pd.to_numeric(df['tripduration'], errors='coerce')

# 6. Fahrtzeit in Minuten berechnen
df['duration_min'] = df['tripduration'] / 60

# 7. Ausgabe prüfen
print("\n🧾 Spaltentypen von tripduration und duration_min:")
print(df[['tripduration', 'duration_min']].dtypes)

print("\n📊 Beispielhafte Einträge:")
print(df[['started_at', 'ended_at', 'tripduration', 'duration_min']].head())
```

```
# Wochentag und Stunde
df['weekday'] = df['started_at'].dt.weekday
df['hour'] = df['started_at'].dt.hour

# Tageszeitkategorie
def get_time_category(hour):
    if 6 <= hour < 12:
        return 'Morgen'
    elif 12 <= hour < 18:
        return 'Nachmittag'
    elif 18 <= hour < 24:
        return 'Abend'
    else:
        return 'Nacht'

df['time_of_day'] = df['hour'].apply(get_time_category)
```
